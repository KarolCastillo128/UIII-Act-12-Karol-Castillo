####  Proyecto de simulación de inteligencia artificial  utiliza POO, tipo Menu PLOMERO

```python
# ==========================================
# SISTEMA INTELIGENTE DE GESTIÓN DE PLOMEROS
# Programación Orientada a Objetos (POO)
# ==========================================

class Plomero:

    def __init__(self, nombre, herramientas, emergencia,
                 materiales, licencia, calificacion, anios):
        self.nombre = nombre
        self.herramientas = herramientas
        self.emergencia = emergencia
        self.materiales = materiales
        self.licencia = licencia
        self.calificacion = calificacion
        self.anios = anios

    # Simulación sencilla de IA
    def recomendacion_ia(self):

        if self.calificacion >= 4.5 and self.anios >= 5:
            return "⭐ Experto Recomendado"

        elif self.emergencia.lower() == "si":
            return "🚨 Disponible Emergencias"

        else:
            return "🔧 Plomero General"


class SistemaPlomeros:

    def __init__(self):
        self.plomeros = []

    def agregar_plomero(self):

        print("\n=== AGREGAR PLOMERO ===")

        nombre = input("Nombre: ")
        herramientas = input("Kit de herramientas: ")
        emergencia = input("Disponibilidad de emergencia (Si/No): ")
        materiales = input("Materiales frecuentes: ")
        licencia = input("Licencia: ")

        calificacion = float(input("Calificación de clientes (0-5): "))
        anios = int(input("Años de servicio: "))

        nuevo = Plomero(
            nombre,
            herramientas,
            emergencia,
            materiales,
            licencia,
            calificacion,
            anios
        )

        self.plomeros.append(nuevo)

        print("\n✅ Plomero agregado correctamente.")

    def listar_plomeros(self):

        print("\n=== LISTADO DE PLOMEROS ===")

        if len(self.plomeros) == 0:
            print("No hay registros.")
            return

        print("-" * 170)

        print(
            f"{'ID':<4}"
            f"{'NOMBRE':<15}"
            f"{'KIT HERRAMIENTAS':<25}"
            f"{'EMERGENCIA':<12}"
            f"{'MATERIALES':<25}"
            f"{'LICENCIA':<15}"
            f"{'CALIF.':<10}"
            f"{'AÑOS':<8}"
            f"{'IA':<30}"
        )

        print("-" * 170)

        for i, p in enumerate(self.plomeros):

            print(
                f"{i:<4}"
                f"{p.nombre:<15}"
                f"{p.herramientas:<25}"
                f"{p.emergencia:<12}"
                f"{p.materiales:<25}"
                f"{p.licencia:<15}"
                f"{p.calificacion:<10}"
                f"{p.anios:<8}"
                f"{p.recomendacion_ia():<30}"
            )

        print("-" * 170)

    def actualizar_plomero(self):

        print("\n=== ACTUALIZAR PLOMERO ===")

        self.listar_plomeros()

        if len(self.plomeros) == 0:
            return

        indice = int(input("Ingrese ID a actualizar: "))

        if indice >= 0 and indice < len(self.plomeros):

            p = self.plomeros[indice]

            p.nombre = input("Nuevo nombre: ")
            p.herramientas = input("Nuevo kit de herramientas: ")
            p.emergencia = input("Disponibilidad emergencia (Si/No): ")
            p.materiales = input("Materiales frecuentes: ")
            p.licencia = input("Licencia: ")
            p.calificacion = float(input("Calificación: "))
            p.anios = int(input("Años de servicio: "))

            print("\n✅ Registro actualizado.")

        else:
            print("ID no válido.")

    def borrar_plomero(self):

        print("\n=== BORRAR PLOMERO ===")

        self.listar_plomeros()

        if len(self.plomeros) == 0:
            return

        indice = int(input("Ingrese ID a eliminar: "))

        if indice >= 0 and indice < len(self.plomeros):

            eliminado = self.plomeros.pop(indice)

            print(f"\n🗑️ {eliminado.nombre} eliminado correctamente.")

        else:
            print("ID no válido.")


# ======================
# PROGRAMA PRINCIPAL
# ======================

sistema = SistemaPlomeros()

opcion = 0

while opcion != 5:

    print("\n")
    print("=" * 40)
    print("   SISTEMA INTELIGENTE DE PLOMEROS")
    print("=" * 40)
    print("1. Agregar plomero")
    print("2. Listar plomeros")
    print("3. Actualizar plomero")
    print("4. Borrar plomero")
    print("5. Salir")
    print("=" * 40)

    opcion = int(input("Seleccione una opción: "))

    if opcion == 1:
        sistema.agregar_plomero()

    elif opcion == 2:
        sistema.listar_plomeros()

    elif opcion == 3:
        sistema.actualizar_plomero()

    elif opcion == 4:
        sistema.borrar_plomero()

    elif opcion == 5:
        print("\n👋 Gracias por utilizar el sistema.")

    else:
        print("\n❌ Opción no válida.")
```
